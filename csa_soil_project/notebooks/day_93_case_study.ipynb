# Day 93: A one-page case study

**Goal:** a professor or recruiter will not open six notebooks. They read **one page**. Today I wrote it: `csa_soil_project/CASE_STUDY.md`.

**The picture:** a news story. The headline and the first line give the news. The details come after.

## 1. Four rules
1. **Lead with the finding, not the steps.** Weak: "I loaded data, merged tables and ran several checks." Strong: "One dataset reports organic carbon 25 to 43 times as high as the others, and I cannot yet say why."
2. **Numbers, not adjectives.** "Very high" tells the reader nothing. "103.4 g/kg against 2.4 to 4.2" does. Every number comes from code.
3. **Say what I do not know.** A reader trusts "the cause is unconfirmed" more than a claim that everything is solved.
4. **Keep it short:** about 300 to 400 words, short sentences, plain words, one voice ("I").

## 2. Exercise (made-up facts)
Facts: three datasets A, B, C with 3 layers each. Medians A = 5, B = 6, C = 110. Same units. The cause is unknown.

**Rewrite of the weak sentence** ("I did a lot of analysis and found some very interesting things in the data"):
> Dataset C has an anomaly with an unknown cause; its median is 22 times that of A and 18 times that of B.

**Four sentences:**
- **Finding:** Dataset C has an organic carbon median that is far higher than the others.
- **Evidence:** The median for C is 110, while A and B are 5 and 6.
- **Limit:** This is based on only 3 layers per dataset.
- **Next step:** collect more layers per dataset before comparing, and check the documentation for units.

**Corrections from review:** cut the word "very" (the number already says it). A next step must name an action: "needs further investigation" is not a next step.

**Self-check list:** finding in sentence 1? at least two numbers? one honest limit? one concrete next step? no "very", "huge" or "amazing"?

**Check question: why does a good case study include what I do not know?**
It builds trust and stops other researchers from using the findings wrongly. If I hide the limits, I look like a salesman. If I admit them, I look like a scientist.
